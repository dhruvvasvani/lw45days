```python
# PDF/Word -> N-up "micro file" converter, interactive, w/ color->B&W option
# Deps: pip install pymupdf pillow
# Word input needs libreoffice installed on system (headless convert)

import fitz  # PyMuPDF
from PIL import Image
import os, subprocess, math, io

def convert_to_pdf_if_needed(path):
    if path.lower().endswith(".pdf"):
        return path
    out_dir = os.path.dirname(path) or "."
    subprocess.run(["libreoffice", "--headless", "--convert-to", "pdf", "--outdir", out_dir, path], check=True)
    base = os.path.splitext(os.path.basename(path))[0]
    return os.path.join(out_dir, base + ".pdf")

def get_grid(n):
    grids = {2: (1, 2), 4: (2, 2), 6: (2, 3), 8: (2, 4)}
    if n not in grids:
        raise ValueError("pages_per_sheet must be 2, 4, 6, or 8")
    return grids[n]

def make_nup(input_path, output_path, pages_per_sheet, bw=False, sheet_size=(842, 595), dpi=150):
    pdf_path = convert_to_pdf_if_needed(input_path)
    src = fitz.open(pdf_path)
    out = fitz.open()

    rows, cols = get_grid(pages_per_sheet)
    sheet_w, sheet_h = sheet_size
    cell_w, cell_h = sheet_w / cols, sheet_h / rows

    total = len(src)
    sheets = math.ceil(total / pages_per_sheet)
    zoom = dpi / 72

    for s in range(sheets):
        page = out.new_page(width=sheet_w, height=sheet_h)
        for cell in range(pages_per_sheet):
            pnum = s * pages_per_sheet + cell
            if pnum >= total:
                break
            r, c = divmod(cell, cols)
            rect = fitz.Rect(c * cell_w, r * cell_h, (c + 1) * cell_w, (r + 1) * cell_h)

            pix = src[pnum].get_pixmap(matrix=fitz.Matrix(zoom, zoom))
            img = Image.open(io.BytesIO(pix.tobytes("png")))
            if bw:
                img = img.convert("L")  # grayscale
            buf = io.BytesIO()
            img.save(buf, format="PNG")
            page.insert_image(rect, stream=buf.getvalue())

    out.save(output_path)
    out.close()
    src.close()
    print(f"Done. {total} pages -> {sheets} sheets, {pages_per_sheet}/sheet, {'B&W' if bw else 'color'} -> {output_path}")

# ---- Run in Jupyter cell (interactive prompts) ----
input_path = input("Input file path (.pdf or .docx): ").strip()
output_path = input("Output file path (e.g. output_nup.pdf): ").strip()
pages_per_sheet = int(input("Pages per sheet (2, 4, 6, 8): ").strip())
bw_choice = input("Convert to B&W? (y/n): ").strip().lower()
bw = bw_choice == "y"

make_nup(input_path, output_path, pages_per_sheet, bw=bw)
```
