```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

projects = [
    {
        "num": "P1", 
        "title": "Business Problem Framing", 
        "data": "train.xlsx, test.xlsx", 
        "skills": "Churn Mapping, Loss Function Translation, KPI Scoping"
    },
    {
        "num": "P2", 
        "title": "Data Ingestion & Cleaning", 
        "data": "Raw Tabular Logs", 
        "skills": "IQR Outlier Removal, Median Imputation, Type Casting"
    },
    {
        "num": "P3", 
        "title": "Exploratory Data Analysis", 
        "data": "User Activity Dataset", 
        "skills": "Correlation Matrix, Skewness Profiling, Visual Histograms"
    },
    {
        "num": "P4", 
        "title": "Feature Engineering", 
        "data": "Analytics Feature Matrix", 
        "skills": "StandardScaler, One-Hot Encoding, Interaction Variables"
    },
    {
        "num": "P5", 
        "title": "Statistical Decision Support", 
        "data": "A/B Experiment Logs", 
        "skills": "Chi-Square Tests, Two-Sample t-tests, 95% Confidence Intervals"
    },
    {
        "num": "P6", 
        "title": "ML Feasibility Assessment", 
        "data": "Preprocessed Baseline Set", 
        "skills": "Dummy Benchmarking, Signal-to-Noise Ratio, Class Balancing"
    },
    {
        "num": "P7", 
        "title": "Predictive Modeling & Tuning", 
        "data": "data.csv", 
        "skills": "Random Forest, GridSearchCV Tuning, Confusion Matrix, ROC-AUC"
    },
    {
        "num": "P8", 
        "title": "Customer Segmentation", 
        "data": "Mall_Customers.csv", 
        "skills": "K-Means Clustering, Elbow Method, RFM Profiling"
    },
    {
        "num": "P9", 
        "title": "AI Use-Case & Governance", 
        "data": None, 
        "focus": "SHAP Explainability, Algorithmic Bias Audit, GDPR Compliance"
    },
    {
        "num": "P10", 
        "title": "Production ML Workflow", 
        "data": None, 
        "focus": "FastAPI REST Endpoint, Docker Container, KS Drift Tracking"
    }
]

fig, ax = plt.subplots(figsize=(16, 12), dpi=300)
fig.patch.set_facecolor('#f8fafc')
ax.set_facecolor('#f8fafc')
ax.axis('off')

plt.text(0.5, 0.96, "DATA SCIENCE & ML PORTFOLIO", color='#0f172a', fontsize=22, fontweight='bold', ha='center')
plt.text(0.5, 0.93, "10-Project End-to-End Applied Machine Learning Journey", color='#475569', fontsize=12, fontweight='medium', ha='center')

cols = 2
card_w, card_h = 0.45, 0.14

for i, p in enumerate(projects):
    col = i % cols
    row = i // cols
    
    x = 0.03 + col * 0.49
    y = 0.75 - row * 0.17
    
    rect = patches.FancyBboxPatch(
        (x, y), card_w, card_h, 
        boxstyle="round,pad=0.01", 
        linewidth=1.2, 
        edgecolor='#cbd5e1', 
        facecolor='#ffffff'
    )
    ax.add_patch(rect)
    
    tag_rect = patches.FancyBboxPatch(
        (x + 0.015, y + 0.092), 0.05, 0.032, 
        boxstyle="round,pad=0.004", 
        facecolor='#0284c7', 
        edgecolor='none'
    )
    ax.add_patch(tag_rect)
    ax.text(x + 0.04, y + 0.108, p["num"], color='white', fontsize=9.5, fontweight='bold', ha='center', va='center')
    
    ax.text(x + 0.075, y + 0.108, p["title"], color='#0369a1', fontsize=11, fontweight='bold', va='center')
    
    if p["data"] is not None:
        ax.text(x + 0.02, y + 0.06, f"Dataset: {p['data']}", color='black', fontsize=9, fontweight='normal')
        ax.text(x + 0.02, y + 0.02, f"Learned: {p['skills']}", color='black', fontsize=8.5, fontweight='normal', style='italic')
    else:
        ax.text(x + 0.02, y + 0.04, f"Learned: {p['focus']}", color='black', fontsize=9, fontweight='normal', style='italic')

plt.savefig("data_science_portfolio_light_clean1.png", bbox_inches='tight', pad_inches=0.3, facecolor=fig.get_facecolor())
plt.show()

print("Clean light infographic successfully saved as 'data_science_portfolio_light_clean1.png'!")
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import matplotlib.patches as patches

plt.style.use('default')
fig = plt.figure(figsize=(20, 24), dpi=300)
fig.patch.set_facecolor('#f1f5f9')

plt.suptitle("DATA SCIENCE & ML PORTFOLIO", fontsize=24, fontweight='bold', y=0.98, color='#0f172a')
plt.figtext(0.5, 0.965, "10-Project Showcase of Applied Data Science", 
            fontsize=13, ha='center', color='#334155', fontweight='medium')

gs = fig.add_gridspec(5, 2, hspace=0.35, wspace=0.20, top=0.94, bottom=0.03, left=0.05, right=0.95)

def style_card_content(ax, title, dataset_text, learned_text):
    ax.set_facecolor('white')
    for spine in ax.spines.values():
        spine.set_edgecolor('#cbd5e1')
        spine.set_linewidth(1.5)
    ax.set_title(title, fontsize=12, fontweight='bold', color='#0369a1', loc='left', pad=10)
    ax.set_xlabel(learned_text, fontsize=9.5, color='black', style='italic', labelpad=5)
    ax.text(0.01, -0.15, dataset_text, color='black', fontsize=9, transform=ax.transAxes, fontweight='normal')

ax1 = fig.add_subplot(gs[0, 0])
ax1.axis('off')
style_card_content(ax1, "P1: Business Problem Framing", 
                   "• Dataset: train.xlsx, test.xlsx", 
                   "Learned: Revenue KPIs to ML classification triggers.")
ax1.text(0.12, 0.65, "Business Pain\nHigh Churn", bbox=dict(boxstyle="round,pad=0.5", fc="#e0f2fe", ec="#0284c7"), color="#0369a1", fontsize=8, ha="center", fontweight='bold')
ax1.text(0.5, 0.65, "ML Translation\nClassification", bbox=dict(boxstyle="round,pad=0.5", fc="#bae6fd", ec="#0284c7"), color="#0369a1", fontsize=8, ha="center", fontweight='bold')
ax1.text(0.88, 0.65, "Impact\n+15% LTV", bbox=dict(boxstyle="round,pad=0.5", fc="#7dd3fc", ec="#0284c7"), color="#0369a1", fontsize=8, ha="center", fontweight='bold')
ax1.annotate('', xy=(0.33, 0.65), xytext=(0.24, 0.65), arrowprops=dict(arrowstyle="->", color="#0284c7", lw=2))
ax1.annotate('', xy=(0.73, 0.65), xytext=(0.64, 0.65), arrowprops=dict(arrowstyle="->", color="#0284c7", lw=2))

ax2 = fig.add_subplot(gs[0, 1])
features = ['Age', 'Income', 'Score']
before = [18, 12, 7]; after = [0, 0, 0]
x = np.arange(len(features))
bars_b = ax2.bar(x - 0.2, before, 0.35, label='Before', color='#f87171')
bars_a = ax2.bar(x + 0.2, after, 0.35, label='After', color='#4ade80')
ax2.set_xticks(x); ax2.set_xticklabels(features, fontsize=8)
ax2.legend(fontsize=7, facecolor='white', edgecolor='none')
ax2.set_ylabel("Null Count", fontsize=8, color='black')
style_card_content(ax2, "P2: Ingestion & Cleaning", 
                   "• Dataset: Raw Tabular Logs", 
                   "Learned: Imputation & IQR outlier handling.")

ax3 = fig.add_subplot(gs[1, 0])
data = np.random.normal(50, 15, 600)
ax3.hist(data, bins=22, color='#38bdf8', edgecolor='#0284c7', alpha=0.7)
ax3.set_ylabel("Frequency", fontsize=8, color='black')
style_card_content(ax3, "P3: Exploratory Data Analysis", 
                   "• Dataset: User Activity Dataset", 
                   "Learned: Statistical distributions & skewness profiling.")

ax4 = fig.add_subplot(gs[1, 1])
raw_feat = np.random.uniform(10, 100, 300)
scaled_feat = (raw_feat - raw_feat.mean()) / raw_feat.std()
sns.kdeplot(scaled_feat, ax=ax4, color='#a855f7', fill=True, alpha=0.2, lw=2)
ax4.set_ylabel("Density", fontsize=8, color='black')
style_card_content(ax4, "P4: Feature Engineering", 
                   "• Methods: StandardScaler & Encoding", 
                   "Learned: Data transformation & standardization.")

ax5 = fig.add_subplot(gs[2, 0])
x = np.linspace(-4, 4, 100)
ax5.plot(x, np.exp(-x**2/2), label='A', color='#64748b')
ax5.plot(x, np.exp(-(x-1.0)**2/2), label='B', color='#0284c7', lw=2)
ax5.fill_between(x, np.exp(-(x-1.0)**2/2), alpha=0.2, color='#0284c7')
ax5.legend(fontsize=7, facecolor='white', edgecolor='none')
style_card_content(ax5, "P5: Statistical Decision Support", 
                   "• Technique: A/B Hypothesis Testing", 
                   "Learned: p-value analysis & statistical significance.")

ax6 = fig.add_subplot(gs[2, 1])
models = ['Dummy', 'LogReg', 'RF']
accs = [0.50, 0.81, 0.93]
bars = ax6.barh(models, accs, color=['#cbd5e1', '#38bdf8', '#0284c7'], height=0.5)
ax6.bar_label(bars, fmt='%.2f', fontsize=8, color='black', padding=3)
ax6.set_xlim(0, 1.15)
style_card_content(ax6, "P6: ML Feasibility Assessment", 
                   "• Benchmark: Dummy Baseline Model", 
                   "Learned: Signal vs Noise benchmark check.")

ax7 = fig.add_subplot(gs[3, 0])
cm = np.array([[70, 2], [1, 41]])
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False, ax=ax7, annot_kws={"size": 10, "color": "black"})
ax7.set_xticklabels(['Pred 0', 'Pred 1'], fontsize=8); ax7.set_yticklabels(['Actual 0', 'Actual 1'], fontsize=8)
style_card_content(ax7, "P7: Predictive Modeling & Tuning", 
                   "• Dataset: data.csv", 
                   "Learned: GridSearchCV hyperparameter optimization.")

ax8 = fig.add_subplot(gs[3, 1])
np.random.seed(42); c1 = np.random.randn(25, 2) + [2, 2]; c2 = np.random.randn(25, 2) + [7, 7]
ax8.scatter(c1[:, 0], c1[:, 1], color='#ef4444', s=25, label='Segment A'); ax8.scatter(c2[:, 0], c2[:, 1], color='#10b981', s=25, label='Segment B')
ax8.legend(fontsize=6.5, loc='upper left', facecolor='white', edgecolor='none')
style_card_content(ax8, "P8: Customer Segmentation", 
                   "• Dataset: Mall_Customers.csv", 
                   "Learned: K-Means clustering via Elbow Method.")

ax9 = fig.add_subplot(gs[4, 0])
ax9.axis('off')
style_card_content(ax9, "P9: AI Design & Governance", 
                   "• Architecture: Design & System Strategy", 
                   "Learned: SHAP explainability, AI Ethics & fairness audit.")
ax9.text(0.12, 0.65, "Bias Audit", bbox=dict(boxstyle="square,pad=0.5", fc="#fef2f2", ec="#ef4444"), color="#991b1b", fontsize=8, ha='center', fontweight='bold')
ax9.text(0.50, 0.65, "Privacy Shield", bbox=dict(boxstyle="square,pad=0.5", fc="#fef3c7", ec="#f59e0b"), color="#92400e", fontsize=8, ha='center', fontweight='bold')
ax9.text(0.88, 0.65, "SHAP", bbox=dict(boxstyle="square,pad=0.5", fc="#f0fdf4", ec="#22c55e"), color="#166534", fontsize=8, ha='center', fontweight='bold')

ax10 = fig.add_subplot(gs[4, 1])
ax10.axis('off')
style_card_content(ax10, "P10: Production ML Workflow", 
                   "• Framework: FastAPI, Docker, MLOps Pipeline", 
                   "Learned: API deployment & automated drift monitoring.")
stages = ["Ingest", "Clean", "Infer API", "Monitor"]
for i, st in enumerate(stages):
    ax10.text(0.12 + i*0.25, 0.6, st, bbox=dict(boxstyle="round,pad=0.5", fc="#f1f5f9", ec="#0284c7"), color="#0369a1", fontsize=7.5, ha="center", fontweight='bold')
    if i < 3:
        ax10.annotate('', xy=(0.24 + i*0.25, 0.6), xytext=(0.18 + i*0.25, 0.6), arrowprops=dict(arrowstyle="->", color="#0284c7", lw=1.5))

plt.savefig("data_science_portfolio_fixed.png", bbox_inches='tight', pad_inches=0.3, facecolor=fig.get_facecolor())
plt.show()

print("Corrected, boxed infographic saved as 'data_science_portfolio_fixed.png'!")
```
