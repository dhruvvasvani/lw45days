## Project 6: ML Feasibility Assessment


```python
# Framework: Evaluating Project Viability & Machine Learning Approaches
```

```python
import pandas as pd

df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

print("Dataset Shape:", df.shape)
df[['customerID', 'tenure', 'Contract', 'MonthlyCharges', 'Churn']].head(3)
```

## Logic & Evaluation

```python
churn_by_contract = df.groupby('Contract')['Churn'].value_counts(normalize=True).unstack()

print("Churn Rate by Contract Type:")
print((churn_by_contract * 100).round(2))

print("\n DECISION: DOES ML MAKE SENSE?")
print("YES. Simple rule-based conditional logic (IF/ELSE) is insufficient because churn triggers are multi-dimensional and non-linear.")
print("Machine Learning makes sense here to capture complex interactions across 20+ customer features.")
```

## suitable ml approach

```python
# Identify ML Approach
target = df['Churn']

print("Target variable:", target.unique())
print("Task type: Supervised Binary Classification")
print("Primary models to try: Logistic Regression (baseline), Random Forest, XGBoost")
```

## Checking dataset size, structure, and labels

```python
print("Rows, Cols:", df.shape)
print("\nData Types:")
print(df.dtypes.value_counts())

print("\nTarget Label Distribution:")
print(df['Churn'].value_counts(normalize=True).round(3))
```

## Justifying feasibility or rejection

```python
n_rows = len(df)
has_labels = 'Churn' in df.columns
total_missing = df.isnull().sum().sum()

is_size_ok = n_rows >= 1000
is_labeled = has_labels
is_clean = total_missing == 0

print("AUTOMATED FEASIBILITY SUMMARY ")
print(f"Data Volume: {n_rows} rows | Status: {'PASS' if is_size_ok else 'FAIL'}")
print(f"Target Labels Available: {has_labels} | Status: {'PASS' if is_labeled else 'FAIL'}")
print(f"Total Missing Values: {total_missing} | Status: {'PASS' if is_clean else 'WARN'}")

if is_size_ok and is_labeled:
    verdict = "FEASIBLE"
    reason = f"Dataset meets volume threshold ({n_rows} > 1,000) and contains valid target labels."
else:
    verdict = "REJECTED"
    reason = "Dataset fails core ML criteria (insufficient rows or missing target column)."

print(f"\nFINAL VERDICT: {verdict}")
print(f"Reason: {reason}")
```
